```python
#Program 1: Employee Dataset Summary

import pandas as pd
import numpy as np

# Sample Employee Dataset
employee = pd.DataFrame({
    'Emp_ID': [101,102,103,104,105],
    'Name': ['Alice','Bob','Charlie','David','Eva'],
    'Age': [25,30,np.nan,28,35],
    'Department': ['HR','IT','Finance','IT','HR'],
    'Salary': [35000,45000,50000,np.nan,60000]
})

print("Employee Dataset")
print(employee)

print("\nNumber of Rows:", employee.shape[0])
print("Number of Columns:", employee.shape[1])

print("\nData Types")
print(employee.dtypes)

print("\nMissing Values")
print(employee.isnull().sum())

print("\nStatistical Summary")
print(employee.describe(include='all'))
```

```python
#Program 2: Frequency Distribution of Students' Grades (Bar Chart)

import pandas as pd
import matplotlib.pyplot as plt

grades = ['A','B','A','C','B','A','B','C','A','D',
          'B','A','C','B','A','A','B','C','D','A']

df = pd.DataFrame({'Grade': grades})

frequency = df['Grade'].value_counts().sort_index()

print(frequency)

frequency.plot(kind='bar')
plt.title("Frequency Distribution of Grades")
plt.xlabel("Grades")
plt.ylabel("Frequency")
plt.show()
```

```python
#Program 3: Histogram of House Prices

import pandas as pd
import matplotlib.pyplot as plt

house_prices = [150000,180000,200000,220000,250000,
                300000,320000,350000,370000,400000,
                450000,500000,750000]

df = pd.DataFrame({'House_Price': house_prices})

plt.hist(df['House_Price'], bins=6, edgecolor='black')
plt.title("Histogram of House Prices")
plt.xlabel("House Price")
plt.ylabel("Frequency")
plt.show()

print("Distribution appears Right Skewed due to high-priced houses.")
```

```python
#Program 5: Detect Outliers Using IQR Method

import pandas as pd

marks = [45,50,55,60,62,65,68,70,72,75,
         78,80,82,85,90,120]

df = pd.DataFrame({'Marks': marks})

Q1 = df['Marks'].quantile(0.25)
Q3 = df['Marks'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['Marks'] < lower) | (df['Marks'] > upper)]

print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)

print("\nOutliers")
print(outliers)
```

```python
#Program 6: Remove Outliers Using IQR

import pandas as pd

sales = [100,120,130,140,150,160,170,
         180,190,200,1000]

df = pd.DataFrame({'Sales': sales})

Q1 = df['Sales'].quantile(0.25)
Q3 = df['Sales'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

cleaned = df[(df['Sales'] >= lower) & (df['Sales'] <= upper)]

print("Original Dataset")
print(df)

print("\nCleaned Dataset")
print(cleaned)
```

```python
#Program 7: Histograms and Box Plots Before & After Removing Outliers

import pandas as pd
import matplotlib.pyplot as plt

data = [20,22,25,27,28,30,32,35,36,38,40,42,150]

df = pd.DataFrame({'Values': data})

Q1 = df['Values'].quantile(0.25)
Q3 = df['Values'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

cleaned = df[(df['Values'] >= lower) & (df['Values'] <= upper)]

# Histogram Before
plt.hist(df['Values'], edgecolor='black')
plt.title("Histogram Before Removing Outliers")
plt.show()

# Histogram After
plt.hist(cleaned['Values'], edgecolor='black')
plt.title("Histogram After Removing Outliers")
plt.show()

# Box Plot Before
plt.boxplot(df['Values'])
plt.title("Box Plot Before Removing Outliers")
plt.show()

# Box Plot After
plt.boxplot(cleaned['Values'])
plt.title("Box Plot After Removing Outliers")
plt.show()
```

```python
#Program 8: Exploratory Data Analysis (EDA) using Iris Dataset

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load Iris Dataset
df = sns.load_dataset('iris')

print("First Five Rows")
print(df.head())

print("\nDataset Information")
print(df.info())

print("\nMissing Values")
print(df.isnull().sum())

print("\nDescriptive Statistics")
print(df.describe())

# Histograms
df.hist(figsize=(8,6))
plt.suptitle("Histograms")
plt.show()

# Box Plots
plt.figure(figsize=(8,6))
sns.boxplot(data=df.iloc[:,0:4])
plt.title("Box Plots")
plt.show()

# Detect Outliers using IQR
numeric = df.select_dtypes(include='number')

Q1 = numeric.quantile(0.25)
Q3 = numeric.quantile(0.75)
IQR = Q3 - Q1

outliers = ((numeric < (Q1 - 1.5*IQR)) |
            (numeric > (Q3 + 1.5*IQR)))

print("\nOutliers in Each Column")
print(outliers.sum())

# Remove Outliers
cleaned = df[~((numeric < (Q1 - 1.5*IQR)) |
               (numeric > (Q3 + 1.5*IQR))).any(axis=1)]

print("\nOriginal Shape:", df.shape)
print("Cleaned Shape:", cleaned.shape)

# Save Cleaned Dataset
cleaned.to_csv("iris_cleaned.csv", index=False)

print("\nCleaned dataset saved as iris_cleaned.csv")
```
